**Daily Expense Tracker**

```python
import numpy as np

def analyze_expenses(expenses_param):
    """Collects daily expenses from the user, stores them, and calculates statistics."""
    day_count = 1
    print("\nEnter your daily expenses. Type 'done' when finished.")

    while True:
        user_input = input(f"Enter expense for day {day_count} (or 'done'):")
        if user_input.lower() == "done":
            break
        try:
            expense = float(user_input)
            if expense < 0:
                print("Expenses cannot be negative. Please enter a positive number.")
            else:
                expenses_param.append(expense)
                day_count += 1
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    if not expenses_param:
        print("No expenses entered. Exiting analysis.")
        return None, None, None, None

    daily_expenses_array = np.array(expenses_param)

    total_expenses = np.sum(daily_expenses_array)
    average_expenses = np.mean(daily_expenses_array)
    max_expenses = np.max(daily_expenses_array)

    # Find the day with the max expense (Numpy returns the first occurrence index)
    # Add 1 because array indices are 0-based, and day_count started from 1
    max_expense_day = np.argmax(daily_expenses_array) + 1

    return total_expenses, average_expenses, max_expenses, max_expense_day

# Main program Execution
if __name__ == "__main__":
    total_expenses, average_expenses, max_expenses, max_expense_day = analyze_expenses([])

    if total_expenses is not None:
        print(f"\n---Expense Summary---")
        print(f"Total Expenses: ${total_expenses:.2f}")
        print(f"Average Expenses: ${average_expenses:.2f}")
        print(f"Maximum Expenses: ${max_expenses:.2f} on Day {max_expense_day}")
        print("---------------------")
    else:
        print("No expenses entered for analysis.")
```
